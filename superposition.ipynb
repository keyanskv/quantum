```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
from qiskit.primitives import Sampler
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt

# Step 1: Create a Quantum Circuit with 1 qubit and 1 classical bit
qc = QuantumCircuit(1, 1)

# Step 2: Apply Hadamard Gate to create superposition
qc.h(0)

# Step 3: Measure the qubit
qc.measure(0, 0)

print("Original Quantum Circuit:")
print(qc)
#qc.draw('mpl')
# Step 4: Transpile for the simulator
simulator = AerSimulator()
transpiled_circuit = transpile(qc, simulator, optimization_level=3)
print("\nTranspiled Quantum Circuit:")
print(transpiled_circuit)
#transpiled_circuit.draw('mpl')
# Step 5: Use Sampler to execute
sampler = Sampler()
job = sampler.run([transpiled_circuit])
result = job.result()

# Step 6: Extract and plot results
counts = result.quasi_dists[0].binary_probabilities()
print("\nMeasurement Results:", counts)
plot_histogram(counts)

```

```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
from qiskit.primitives import Sampler
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt

# Step 1: Create a Quantum Circuit with 2 qubits and 2 classical bits
qc = QuantumCircuit(2, 2)

# Step 2: Apply a Hadamard gate to the first qubit to create superposition
qc.h(0)

# Step 3: Apply a CNOT gate (Controlled-X) to entangle the qubits
qc.cx(0, 1)

# Step 4: Measure both qubits
qc.measure([0, 1], [0, 1])

# Visualize the circuit
print("Quantum Circuit:")
print(qc.draw('text'))

# Step 5: Simulate using Aer simulator
simulator = AerSimulator()
transpiled_circuit = transpile(qc, simulator, optimization_level=3)
job = simulator.run(transpiled_circuit, shots=1000)
result = job.result()
# Step 6: Display the results
counts = result.get_counts()
print("Results of Bell State Measurement:", counts)

# Plot a histogram of the results
plot_histogram(counts)
```

```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt

# Step 1: Create a Quantum Circuit with 2 qubits and 2 classical bits
qc = QuantumCircuit(2, 2)

# Step 2: Apply X gate to the first qubit to set it to |1⟩
qc.initialize([0, 1], 1)

# Step 3: Apply a Hadamard gate to create superposition on qubit 0
qc.h(0)

# Step 4: Apply a CNOT gate (Controlled-X) to entangle the qubits
qc.cx(0, 1)

# Step 5: Measure both qubits
qc.measure([0, 1], [0, 1])

# Visualize the circuit
print("Quantum Circuit:")
print(qc.draw('text'))

# Step 6: Simulate using AerSimulator
simulator = AerSimulator()
transpiled_circuit = transpile(qc, simulator, optimization_level=3)
job = simulator.run(transpiled_circuit, shots=1000)
result = job.result()

# Step 7: Display the results
counts = result.get_counts()
print("Results of Bell State Measurement:", counts)

# Plot a histogram of the results
plot_histogram(counts)
```
